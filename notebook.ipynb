<div style="background-color: #e7f3ff; padding: 10px; border-left: 4px solid #2196F3; margin: 10px 0;">
  <strong>💡 Qu'est-ce que <code>Stratify Col</code> ?</strong><br>
  Lorsqu'on divise un dataset (ex: auto_split dans [S01](#S01) ou split dans [S07](#S07)), le paramètre <strong>Stratify Col</strong> (Colonne de stratification) force la division à conserver exactement la même distribution de catégories que dans le dataset d'origine pour la colonne choisie (en général, la variable cible). Cela évite, par exemple, qu'une classe minoritaire soit sur/sous-représentée (ou absente !) dans le set de validation/test.
</div>
