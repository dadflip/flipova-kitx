### S07 — Dataset Balancing (Split & Équilibrage)
Équilibrage des classes si nécessaire et division des données en sous-ensembles d'entraînement et de test.

> 💡 **Qu'est-ce que `Stratify Col` ?**<br>
> Lorsqu'on divise un dataset (ex: auto_split dans S01 ou split dans S07), le paramètre **Stratify Col** (Colonne de stratification) force la division à conserver exactement la même distribution de catégories que dans le dataset d'origine pour la colonne choisie (en général, la variable cible). Cela évite, par exemple, qu'une classe minoritaire soit sur/sous-représentée (ou absente !) dans le set de validation/test.
